## 通信时间统计
### comm_kernel_info是所有在GPU设备上的通信时间

```python
comm_kernel_info = analyzer.get_launch_comm_kernels_with_user_annotations(rank=rank, no_overlap=False)
comm_kernel_info.groupby(['tid_cpu','s_user_annotation_cpu']).agg(
    dur_gpu_sum=('dur_gpu', lambda x: x.sum() / 1000),
    count=('dur_gpu', 'count')
).groupby('tid_cpu').apply(lambda x: x.sort_values(by='dur_gpu_sum', ascending=False))
```

### comm_kernel_info_not_overlap是统计没有被计算掩盖的通信时间分布情况

```python
comm_kernel_info_not_overlap = analyzer.get_launch_comm_kernels_with_user_annotations(rank=rank, no_overlap=True)
comm_kernel_info_not_overlap.groupby(['tid_cpu','s_user_annotation_cpu']).agg(
    dur_gpu_sum=('dur_gpu', lambda x: x.sum() / 1000),
    count=('dur_gpu', 'count')
).groupby('tid_cpu').apply(lambda x: x.sort_values(by='dur_gpu_sum', ascending=False))
```
